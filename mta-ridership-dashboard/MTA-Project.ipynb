```python
import pandas as pd

df = pd.read_csv("data/mta_ridership_clean.csv")

# Make sure timestamp is datetime
df["transit_timestamp"] = pd.to_datetime(df["transit_timestamp"])

# Add missing columns
df["date"] = df["transit_timestamp"].dt.date
df["year"] = df["transit_timestamp"].dt.year
df["month"] = df["transit_timestamp"].dt.month
df["day"] = df["transit_timestamp"].dt.day
df["hour"] = df["transit_timestamp"].dt.hour
df["day_of_week"] = df["transit_timestamp"].dt.day_name()
df["is_weekend"] = df["day_of_week"].isin(["Saturday", "Sunday"])

# Make sure numeric fields are correct
df["ridership"] = pd.to_numeric(df["ridership"], errors="coerce")
df["transfers"] = pd.to_numeric(df["transfers"], errors="coerce")

# Save OVER the old CSV
df.to_csv("data/mta_ridership_clean.csv", index=False)

print(df.columns)
print("CSV updated successfully")
```

```python
print(df.head())
print(df.shape)
```

```python
top_stations = (
    df.groupby("station_complex")["ridership"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_stations)
```

```python
hourly_ridership = (
    df.groupby("hour")["ridership"]
    .sum()
    .sort_values()
)

print(hourly_ridership)
```

```python
borough_ridership = (
    df.groupby("borough")["ridership"]
    .sum()
    .sort_values(ascending=False)
)

print(borough_ridership)
```

```python
weekend_comparison = (
    df.groupby("is_weekend")["ridership"]
    .sum()
)

print(weekend_comparison)
```

```python
payment_analysis = (
    df.groupby("payment_method")["ridership"]
    .sum()
    .sort_values(ascending=False)
)

print(payment_analysis)
```
